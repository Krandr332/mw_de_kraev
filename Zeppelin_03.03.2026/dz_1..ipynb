```scala
// Соберите витрину mart_city_top_products, для этого необходимо:
// 1. Посчитать revenue = qty * price
// 2. Объединить orders с users и products
// 3. Посчитать метрики по (city, product_id, product_name):
//  orders_cnt (кол-во заказов)
//  qty_sum (сумма qty)
//  revenue_sum (сумма revenue)э
// 4. Для каждого города выбрать Top-2 товара по revenue_sum используя
// Window
// 5. Записать результат в s3 и в HDFS по пути :
// /tmp/sandbox_zeppelin/mart_city_top_products/ (parquet, overwrite)
// 6. Прочитать обратно и показать топ-результат.
```

```scala
%spark.pyspark

from pyspark.sql import functions as F
users = spark.createDataFrame(
[ ("u1", "Berlin"),
("u2", "Berlin"),
("u3", "Munich"),
("u4", "Hamburg"), ],
["user_id", "city"] )
orders = spark.createDataFrame(
[ ("o1", "u1", "p1", 2, 10.0),
("o2", "u1", "p2", 1, 30.0),
("o3", "u2", "p1", 1, 10.0),
("o4", "u2", "p3", 5, 7.0),
("o5", "u3", "p2", 3, 30.0),
("o6", "u3", "p3", 1, 7.0),
("o7", "u4", "p1", 10, 10.0), ],
["order_id", "user_id", "product_id", "qty", "price"] )
products = spark.createDataFrame(
[ ("p1", "Ring VOLA"),
("p2", "Ring POROG"),
("p3", "Ring TISHINA"), ],
["product_id", "product_name"] )
users.show() 
orders.show()
products.show()
```

```scala
%spark.pyspark

ord_with_rev = orders.withColumn("revenue", F.col("qty") * F.col("price"))
ord_with_rev.show()
```

```scala
%spark.pyspark
ord_enri = ord_with_rev.join(users, "user_id").join(products, "product_id")
ord_enri.show()
```

```scala
%spark.pyspark

city_product_st = ord_enri.groupBy(
    "city", "product_id", "product_name"
).agg(
    F.count("order_id").alias("orders_cnt"),
    F.sum("qty").alias("qty_sum"),
    F.sum("revenue").alias("revenue_sum")
)
city_product_st.show()
```

```scala
%spark.pyspark

from pyspark.sql.window import Window

window_spec = Window.partitionBy("city").orderBy(F.col("revenue_sum").desc())
city_product_st = city_product_st.withColumn(
    "time_top", F.row_number().over(window_spec))
top_products = city_product_st.filter(F.col("time_top") <= 2).drop("time_top")
top_products.show()
```

```scala
%spark.pyspark

output_path = "/tmp/sandbox_zeppelin/mart_city_top_products/"
top_products.write.mode("overwrite").parquet(output_path)
```

```scala
%spark.pyspark

result = spark.read.parquet(output_path)
result.show()
```
